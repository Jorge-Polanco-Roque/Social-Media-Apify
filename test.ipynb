## Extracción de URLs

```python
url_1 = pd.read_csv('Extracción/tiktok_01_task.csv')
```

```python
# Extraer valores únicos de una columna
unique_url_1 = url_1['text'].unique()

# Guardar en txt
with open('tiktok_unique_url.txt', 'w', encoding='utf-8') as f:
    for value in unique_url_1:
        f.write(f"{value}\n")

print(f"Guardados {len(unique_url_1)} valores únicos en valores_unicos.txt")
```
